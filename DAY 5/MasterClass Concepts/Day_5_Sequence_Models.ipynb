### 🧠 The "Memory Brain" Framework
Unlike standard Feed-Forward networks, Sequence Models have a **Recurrent** connection that allows information to persist.

1. **Short-Term Memory (RNN):** Uses a hidden state to carry information. However, it suffers from the **Vanishing Gradient** problem—it "forgets" the beginning of a long sentence.
2. **Long-Term Memory (LSTM):** Introduces a **Cell State** (the conveyor belt) and **Gates** (Forget, Input, Output) to surgically decide what to keep.
3. **Efficient Memory (GRU):** A streamlined version that merges gates for faster training without losing the "Memory" edge.



---

### 🔹 Why Sequence Models?
In NLP, the meaning of a word depends on the words before it.
* *Example:* "I am going to **bank** the money" vs "I am sitting on the river **bank**."
* Context is everything.

```python
# ── HKS SEQUENCE TOOLBOX ───────────────────────────────────────────────
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Global Plotting Style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (12, 5)

# 1. Data Acquisition (Top 10,000 frequent words)
vocab_size = 10000
max_len = 200 # Fixed sequence length for uniformity
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=vocab_size)

# 2. Sequence Padding (Ensuring all reviews are 200 words long)
X_train = pad_sequences(X_train, maxlen=max_len)
X_test = pad_sequences(X_test, maxlen=max_len)

print(f"✅ HKS Engine: Data Loaded. Training Samples: {len(X_train)} | Sequence Length: {max_len}")
```

```python
# ── ARCHITECTURE BENCHMARKING: RNN vs LSTM vs GRU ─────────────────────

def build_model(layer_type, name):
    model = keras.Sequential([
        keras.layers.Embedding(vocab_size, 128, input_length=max_len),
        layer_type(64, dropout=0.2, recurrent_dropout=0.2), # Adding Dropout for HKS Reliability
        keras.layers.Dense(1, activation='sigmoid')
    ], name=name)
    model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
    return model

# Initializing Models
model_rnn  = build_model(keras.layers.SimpleRNN, "HKS_Simple_RNN")
model_lstm = build_model(keras.layers.LSTM, "HKS_Long_Short_Memory")
model_gru  = build_model(keras.layers.GRU, "HKS_Gated_Recurrent_Unit")

# Training (Using 5 Epochs as per your requirement)
print("🚀 Training HKS-RNN...")
history_rnn = model_rnn.fit(X_train, y_train, epochs=5, batch_size=64, validation_data=(X_test, y_test), verbose=1)

print("\n🚀 Training HKS-LSTM...")
history_lstm = model_lstm.fit(X_train, y_train, epochs=5, batch_size=64, validation_data=(X_test, y_test), verbose=1)

print("\n🚀 Training HKS-GRU...")
history_gru = model_gru.fit(X_train, y_train, epochs=5, batch_size=64, validation_data=(X_test, y_test), verbose=1)
```

```python
# ── HKS RESEARCH ANALYTICS: MODEL COMPARISON ──────────────────────────
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(20, 7))

# Plot 1: Validation Accuracy
ax1.plot(history_rnn.history['val_accuracy'], label='Simple RNN', color='#e74c3c', lw=3)
ax1.plot(history_lstm.history['val_accuracy'], label='LSTM', color='#2ecc71', lw=3)
ax1.plot(history_gru.history['val_accuracy'], label='GRU', color='#3498db', lw=3)
ax1.set_title("HKS Research: Accuracy Progression", fontsize=16)
ax1.set_xlabel("Epochs"); ax1.set_ylabel("Validation Accuracy"); ax1.legend()

# Plot 2: Validation Loss
ax2.plot(history_rnn.history['val_loss'], label='Simple RNN', color='#e74c3c', linestyle='--')
ax2.plot(history_lstm.history['val_loss'], label='LSTM', color='#2ecc71', linestyle='--')
ax2.plot(history_gru.history['val_loss'], label='GRU', color='#3498db', linestyle='--')
ax2.set_title("HKS Research: Loss Convergence", fontsize=16)
ax2.set_xlabel("Epochs"); ax2.set_ylabel("Validation Loss"); ax2.legend()

plt.suptitle("Comparative Analysis of Temporal Models", fontsize=20, y=1.05)
plt.show()
```

```python
# ── HKS PREDICTION ENGINE ──
sample_idx = 0
sample = X_test[sample_idx].reshape(1, -1)
actual_label = "Positive" if y_test[sample_idx] == 1 else "Negative"

pred = model_lstm.predict(sample, verbose=0)
pred_label = "Positive" if pred > 0.5 else "Negative"

print(f"📄 Specimen Analysis:")
print(f"   - Actual Sentiment: {actual_label}")
print(f"   - HKS Model Prediction: {pred_label} (Confidence: {pred[0][0]:.4f})")
```

# 🧪 Day 05: Technical Summary & Findings

| Model Architecture | Sequential Speed | Val Accuracy | Memory Depth | HKS Verdict |
| :--- | :--- | :--- | :--- | :--- |
| **Simple RNN** | 🔥 Fastest | ~80% | Weak (Vanishing Gradient) | Good for small sequences only. |
| **LSTM** | 🐢 Slower | **~87%** | Strong (Triple Gating) | Best for complex context & long text. |
| **GRU** | ⚡ Fast | ~86.5% | Efficient (Update/Reset) | Best balance of speed vs performance. |

### 🧬 Key Takeaways:
1. **Vanishing Gradient:** RNNs struggle with long sentences because the error signal dies out over time steps.
2. **Gating Logic:** LSTMs and GRUs use math gates to decide which words to "forget," making them much smarter for NLP.
3. **Embedding Power:** Converting words into 128-dimensional vectors allowed the models to understand the *meaning* behind words, not just their frequency.
